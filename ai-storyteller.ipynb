## 1. Configuration: Ollama + Llama Model

```python
import os
from langchain_ollama import ChatOllama
from langchain.prompts import PromptTemplate

# LLM used for analysis steps 1 to 4
llm_analyst = ChatOllama(model="llama3.2", temperature=0, num_ctx = 24576)
# LLM used for storytelling step 5
llm_storyteller = ChatOllama(model="llama3.2", temperature=0.7, num_ctx = 24576)
```

### Phase 1: Customer Archetype Analyst

```python
PHASE_1_CUSTOMER_ARCHETYPE_ANALYST_PROMPT_TEMPLATE = PromptTemplate(
    input_variables=[
        "first_name",
        "age",
        "gender",
        "professional_background",
        "geographic_location",
        "purchase_history",
        "aov",
        "style_preferences",
        "loyalty_level",
    ],
    template="""
You are an archetype analyst specializing in customer segmentation for personalized luxury marketing strategies.

Analyze the following customer data to identify the most relevant archetype (Purist, Activist, or Expressionist) and provide a short description of the archetype based on the customer’s traits.

**Archetype Descriptions:**

**Purists: The Traditional Connoisseurs**  
Purists represent the older demographic, with 50% aged over 35, and are deeply rooted in tradition and authenticity. They prioritize durable products and authenticity, valuing heritage and brand storytelling in their luxury purchases. This group shows the highest interest in investment pieces, with 36% purchasing a handbag over $500 and 31% buying a luxury timepiece over $500 in the last year.

When it comes to AI, Purists are open to its potential to improve marketing communications, with 38% stating AI would enhance their attitude towards a brand. However, they are selective in their willingness to pay extra for AI-driven enhancements (26% willing to pay 20% more). For Purists, AI is seen as a tool that can help improve efficiency and support people in their jobs, rather than redefining their experience.

**Activists: The Conscious Consumers**  
Activists are younger consumers, with 66% under the age of 35, including Gen Z and younger Millennials. They prioritize high-quality products and sustainability, reflecting their strong alignment with social responsibility. Key values include sustainability and social responsibility (79%) and diversity, equity, and inclusion (76%), making them highly conscious of the ethics and impact of the brands they support.

This group is particularly optimistic about the role of AI in advancing sustainability, with 31% believing it can create more sustainable garments. Additionally, 36% are willing to spend at least 20% more on brands using AI for sustainable fabric development. Activists demonstrate trust in AI (28%) and show excitement about personalized experiences, with a quarter of them stating AI-driven personalization would increase loyalty. Brands leveraging AI to meet sustainability goals and personalized engagement will resonate strongly with this group.

**Expressionists: The Style Individualists**  
Expressionists are millennials aged 25-44 (55%), driven by a desire for self-expression and standing out from the crowd. They prioritize brands that fit their wardrobe and style, with 80% valuing brands that allow them to express their individuality. Expressionists are the most likely to purchase apparel and shoes, with 96% of this group engaging in these categories.

This archetype is highly enthusiastic about AI's potential to enhance creativity, particularly in marketing and design. 82% of Expressionists are willing to pay more for brands that create engaging marketing using generative AI, and 30% would spend at least 20% more if AI is used for better design. Their willingness to pay extra (29%) for AI-driven innovations reflects their openness to unique and improved designs. For Expressionists, AI is seen as a complement to designers, enhancing creativity and aligning with their love for bold, exciting fashion.

Customer Data: 
- First Name: {first_name}  
- Age: {age} 
- Gender: {gender}
- Professional Background: {professional_background} 
- Geographic Location: {geographic_location}
- Purchase History (24 months): {purchase_history}  
- Average Order Value: {aov}
- Style Preferences: {style_preferences} 
- Loyalty Level: {loyalty_level}  

TASK:
Using the provided customer data, identify the customer’s archetype and return a single string with the archetype name and a concise description of how their traits align with that archetype.

EXPECTED OUTPUT:
- Archetype Name: The name of the archetype (Purist, Activist, or Expressionist).  
- Archetype Description: A brief explanation connecting the customer’s traits to the identified archetype.
- Return the output in this format: `Archetype Name: [Archetype Description]`

FORMAT RULES:
- Be precise and data-driven.  
- Avoid subjective judgments.  
- Use only the provided information to make your assessment.  
- Ensure the output aligns with the customer's data and the archetype descriptions provided above.
"""
)
```

```python
def generate_phase_1_output(first_name: str,
                            age: int,
                            gender: str,
                            professional_background: str,
                            geographic_location: str,
                            purchase_history: str,
                            aov: float,
                            style_preferences: str,
                            loyalty_level: str) -> str:
    """
    Generates Phase 1 analysis using the relevant template and input data.
    """
    prompt = PHASE_1_CUSTOMER_ARCHETYPE_ANALYST_PROMPT_TEMPLATE.format(
        first_name=first_name,
        age=age,
        gender=gender,
        professional_background=professional_background,
        geographic_location=geographic_location,
        purchase_history=purchase_history,
        aov=aov,
        style_preferences=style_preferences,
        loyalty_level=loyalty_level
    )
    print("=== Phase 1 Input ===")
    print(prompt)
    return llm_analyst.invoke(prompt)
```

```python
# Generate Phase 1 Output - Archetype
phase_1_result = generate_phase_1_output(
    first_name=first_name_data,
    age=age_data,
    gender=gender_data,
    professional_background=professional_background_data,
    geographic_location=geographic_location_data,
    purchase_history=purchase_history_data,
    aov=average_order_value_data,
    style_preferences=style_preferences_data,
    loyalty_level=loyalty_level_data
)

print("=== Phase 1 Output ===")
print(phase_1_result.content)
```

### Phase 2: Customer Data Analyst

```python
PHASE_2_CUSTOMER_DATA_ANALYST_PROMPT_TEMPLATE = PromptTemplate(
    input_variables=[
        "first_name",
        "age",
        "gender",
        "professional_background",
        "geographic_location",
        "archetype",
        "purchase_history",
        "aov",
        "style_preferences",
        "loyalty_level",
    ],
    template="""
You are a luxury retail data analyst.

Analyze the following customer data to create a comprehensive profile for personalized luxury product narratives:

Customer Data:
- First Name: {first_name}
- Age: {age}
- Gender: {gender}
- Professional Background: {professional_background}
- Geographic Location: {geographic_location}
- Archetype: {archetype}
- Purchase History (24 months): {purchase_history}
- Average Order Value: {aov}
- Style Preferences: {style_preferences}
- Loyalty Level: {loyalty_level}

TASK:
Analyze these data points to create a comprehensive profile for personalized luxury product narratives.
DO NOT add further recommendations or conclusions. Stick your answer to the expected output.

EXPECTED OUTPUT:
1. Customer Demographic Profile
2. Purchase Behavior Analysis
3. Lifestyle Integration Points
4. Cultural Interest Map

FORMAT RULES:
- Be precise and analytical
- Avoid subjective judgments
- Use data-driven insights
- Maintain client privacy
- Focus on actionable insights
"""
)
```

```python
def generate_phase_2_output(first_name: str,
                            age: int,
                            gender: str,
                            professional_background: str,
                            geographic_location: str,
                            archetype: str,
                            purchase_history: str,
                            aov: float,
                            style_preferences: str,
                            loyalty_level: str) -> str:
    """
    Generates Phase 2 analysis using the relevant template and input data.
    """
    prompt = PHASE_2_CUSTOMER_DATA_ANALYST_PROMPT_TEMPLATE.format(
        first_name=first_name,
        age=age,
        gender=gender,
        professional_background=professional_background,
        geographic_location=geographic_location,
        archetype=archetype,
        purchase_history=purchase_history,
        aov=aov,
        style_preferences=style_preferences,
        loyalty_level=loyalty_level
    )
    print("=== Phase 2 Input ===")
    print(prompt)
    return llm_analyst.invoke(prompt)
```

```python
# Generate Phase 2 Output - Customer Profile
phase_2_result = generate_phase_2_output(
    first_name=first_name_data,
    age=age_data,
    gender=gender_data,
    professional_background=professional_background_data,
    geographic_location=geographic_location_data,
    archetype=phase_1_result.content, #chaining phase 1 and phase 2
    purchase_history=purchase_history_data,
    aov=average_order_value_data,
    style_preferences=style_preferences_data,
    loyalty_level=loyalty_level_data
)

print("=== Phase 2 Output ===")
print(phase_2_result.content)
```

### Phase 3: Brand Voice Expert for The Row

```python
PHASE_3_BRAND_VOICE_EXPERT_PROMPT_TEMPLATE = PromptTemplate(
    input_variables=["customer_profile"],
    template="""
You are a branding expert tasked with creating communication materials for The Row, a luxury fashion brand. 
Use The Row's brand guidelines to develop a minimalist aesthetic and maintain understated luxury in language and storytelling. 
The brand emphasizes timeless design, quality craftsmanship, discretion, and intellectual sophistication.

### GUIDELINES
**Tone & Style:**
- **Tone:** Sophisticated, Minimalist, Intellectual, Timeless.
- **Style:** Concise, elegant, and textural. Avoid trendy jargon or excessive superlatives.

**Brand Personality:**
- Private & confident, appealing to a cultured and discerning clientele.
- Sophisticated, intellectual, and timeless.

**Key Messaging Pillars:**
1. Highlight quality and craftsmanship (e.g., exotic skins, 900g cashmere).
2. Emphasize exclusivity and discretion (e.g., "for those who value quiet luxury").
3. Showcase cultural and artistic roots (e.g., references to art, architecture, and design).
4. Convey sensual experiences through tactile descriptions (e.g., "a fabric that feels like a second skin").
5. Promote timelessness and sustainability (e.g., "pieces designed to endure and transcend trends").

**Do’s:**
- Use elegant and refined vocabulary.
- Focus on sensory and tactile descriptions.
- Incorporate cultural references subtly.
- Maintain a minimalist and understated approach.

**Don’ts:**
- Avoid boastful or overly promotional language.
- Refrain from using trendy expressions or clichés.
- Do not focus on discounts or sales; emphasize value and longevity instead.

TASK:
Use the brand guidelines above to create a communication system for **The Row** based on the provided customer profile:  

Customer Profile Analysis:
{customer_profile}

DO NOT add further recommendations or conclusions. Stick your answer to the expected output.

EXPECTED OUTPUT:
1. Approved Vocabulary List
2. Tone Parameters
3. Storytelling Framework
4. Storytelling Pillars

"""
)
```

```python
def generate_phase_3_output(customer_profile: str) -> str:
    """
    Phase 3: Convert the profile into The Row's brand voice.
    """
    prompt = PHASE_3_BRAND_VOICE_EXPERT_PROMPT_TEMPLATE.format(customer_profile=customer_profile)
    print("==== PHASE 3 INPUT ====")
    print(prompt)
    return llm_analyst.invoke(prompt)
```

```python
# Generate Phase 3 Output - Brand Voice
phase_3_result = generate_phase_3_output(
    customer_profile=phase_2_result.content #chaining phase 2 and phase 3
)

print("=== Phase 3 Output ===")
print(phase_3_result.content)
```

### Phase 4: Art and Culture Curator for The Row

```python
PHASE_4_ART_CULTURE_CURATOR_PROMPT_TEMPLATE = PromptTemplate(
    input_variables=["cultural_interests", "product_design_elements"],
    template="""
You are an art and culture curator specializing in luxury fashion contexts. 
Your expertise lies in identifying connections between art movements, paintings, sculptures, architecture, and historical design references to craft culturally rich narratives for customers.

Customer's Cultural Interest Map:
{customer_profile}

Customer Location:
{customer_location}

Product Title:
{product_title}

Product Description:
{product_description}

TASK:
Given the customer’s location, cultural interests, and a product description, identify relevant connections to local art movements, paintings, sculptures, architecture, music and historical design.
Your focus is to create a culturally resonant bridge between the customer, customer location and the product.

DO NOT add further recommendations or conclusions. Stick your answer to the expected output.
DO NOT reference competitors nor other fashion brands.


EXPECTED OUTPUT:
1. Artistic Reference Matrix.
2. Architectural Parallels and Landmarks.
3. Material Culture Connections.
4. Historical Design Links.
5. Contemporary Touchpoints.
6. Music playlist.


CURATION RULES:
- Focus on minimalist movements
- Prioritize timeless references
- Avoid obvious connections
- Maintain subtle sophistication
"""
)
```

```python
def generate_phase_4_output(customer_profile: str,
                            customer_location: str,
                            product_title: str,
                            product_description: str) -> str:
    """
    Phase 4: Add cultural/artistic references.
    """
    prompt = PHASE_4_ART_CULTURE_CURATOR_PROMPT_TEMPLATE.format(
        customer_profile=customer_profile,
        customer_location=customer_location,
        product_title=product_title,
        product_description=product_description
    )
    print("==== PHASE 4 INPUT ====")
    print(prompt)
    return llm_analyst.invoke(prompt)
```

```python
# Generate Phase 4 Output - Art Curator
phase_4_result = generate_phase_4_output(
    customer_profile=phase_2_result.content,
    customer_location=geographic_location_data,
    product_title=product_title_data,
    product_description=product_description_data
)

print("=== Phase 4 Output ===")
print(phase_4_result.content)
```
